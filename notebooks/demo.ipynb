# coordconvertergc Demo

This notebook demonstrates how to install and use the `coordconvertergc` Python package from PyPI.

The package converts geographic coordinates between degrees, minutes, seconds (DMS) and decimal degrees using `to_decimal()` and `to_degree()`.

```python
# install coordconvertergc==0.1.0
%pip install --upgrade --force-reinstall coordconvertergc==0.1.0
```

```python
# Import the two coordinate conversion functions
from coordconvertergc import to_decimal, to_degree
```

```python
# Convert DMS coordinates to decimal degrees
latitude_dms = (15, 47, 0, "S")
longitude_dms = (35, 0, 0, "E")

decimal_coordinates = to_decimal(latitude_dms, longitude_dms)
print("Decimal degrees:", decimal_coordinates)

# Convert decimal degrees back to DMS
dms_coordinates = to_degree(*decimal_coordinates)
print("DMS coordinates:", dms_coordinates)
```

# Expected outputs
Decimal degrees: (-15.783333333333333, 35.0)
DMS coordinates: ((15, 47, 0.0, 'S'), (35, 0, 0.0, 'E'))
